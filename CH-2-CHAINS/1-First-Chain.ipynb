# **CHAINS in langchain**

```python
import os
from dotenv import load_dotenv

load_dotenv()

if os.environ.get("OPENAI_API_KEY"):
    print("Bro API Key Exists")
else:
    raise ValueError("OPENAI_API_KEY not found")

from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_openai import ChatOpenAI
```

# First CHAIN

```python
# TASK -1[Prompt]


prompt_template = ChatPromptTemplate.from_messages([
    ("system", "You are Helpful Assistant"),
    ("human","{input}")
])
```

```python
# TASK-2 [LLM]
llm_openai = ChatOpenAI(model='gpt-5-mini', temperature=0)
```

```python
# TASK-3 String parser
str_parser = StrOutputParser()
```

# Manual Invocation

```python
template = prompt_template.invoke({"input": "What is the capital city of France"})

res = llm_openai.invoke(template)

res.content
```

# Chain Invocation

```python
chain = prompt_template | llm_openai | str_parser

chain.invoke({"input":"what is the capital city of India?"})
```

```python
from langchain_core.runnables import RunnableSequence

chain_1 = RunnableSequence(prompt_template,   llm_openai,   str_parser)
chain_1.invoke({"input":"what is the capital city of USA?"})
```
