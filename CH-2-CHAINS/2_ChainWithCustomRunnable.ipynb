```python
import os
from dotenv import load_dotenv

load_dotenv()

if os.environ.get("OPENAI_API_KEY"):
    print("Bro API Key Exists")
else:
    raise ValueError("OPENAI_API_KEY not found")

from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_openai import ChatOpenAI
from langchain_core.runnables import RunnableLambda
from langchain_core.prompts import PromptTemplate
```

# Chain with Custom Runnable

```python
# TASK -1[Prompt]


prompt_template = ChatPromptTemplate.from_messages([
    ("system", "You are Helpful Assistant"),
    ("human","{input}")
])
```

```python

# TASK-2 [LLM]
llm_openai = ChatOpenAI(model='gpt-5-mini', temperature=0)

```

```python
# TASK-3 String parser
str_parser = StrOutputParser()

```

```python

# TASK -4[Custom Runnable]

def dictionary_maker(text:str) -> dict:
    return {"text": text}

dictionary_maker_runnable = RunnableLambda(dictionary_maker) 
```

```python
# TASK -5 [Template for post]

prompt_post = ChatPromptTemplate.from_messages(
    messages=[
        ("system", "You are a social media post generator"),
        ("human", "Create a post for the following text for Instagram: {text}")
    ]
)
```

```python
# TASK -6 [Generate post with llm]

llm_openai = ChatOpenAI(model="gpt-5-mini", temperature=0)
```

```python
# TASK - 7 [Str parser]

str_parser = StrOutputParser()
```

```python
# CHAIN

chain = prompt_template | llm_openai | str_parser | dictionary_maker_runnable | prompt_post | llm_openai | str_parser

chain.invoke("What is the capital of Canada?")
```
