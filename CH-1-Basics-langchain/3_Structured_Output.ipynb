# **Guiding in Prompt**

```python
from langchain_core.prompts import PromptTemplate
from langchain_openai import ChatOpenAI

llm_openai = ChatOpenAI(model="gpt-5-mini", temperature=0)
result = llm_openai.invoke("Tell me a joke.Generated output in key-value pair format is setup, puchline\n")
result.content
```

# **Using Pydantic Models**

```python
from pydantic import BaseModel, Field

class JokeSchema(BaseModel):
    setup: str = Field(description="The set up for the joke")
    punchline: str = Field(description="The punchline for the joke")

structured_llm = llm_openai.with_structured_output(JokeSchema)
response = structured_llm.invoke("Tell me a joke.")
response
```

# **Using TypedDict Models**

```python
from typing import TypedDict

class llm_schema_td(TypedDict):
    setup: str
    punchline: str
```

```python
obj = llm_schema_td({"ketchup": "Some Setup", "punchline": "Some punchline"})
obj
```

```python
llm_strctured_typed_dict = llm_openai.with_structured_output(llm_schema_td)
response = llm_strctured_typed_dict.invoke("Tell me a joke.")
response
```
